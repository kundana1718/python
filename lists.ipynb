```python
#list in python
#list is an ordered and changeagle collection that can store multiple values
marks=[80,90,75,85]

print(marks)

```

```python
#accessing elements in a list
marks=[80,90,75,85]

print(marks[0])
print(marks[1])
print(marks[2])
print(marks[3])
```

```python
#change element in a list
marks=[80,90,75]

marks[1]=95

print(marks)
```

```python
#change elements in a list
marks=[79,64,92,74,96,77,94]

marks[1]= 95

marks[3]= 80

marks[5]=70
print(marks)
```

```python
#add elements to a list
marks=[80,90,75]

marks.append(85)

print(marks)
```

```python
#add elements to a list
marks=[90,87,56,27,50]

marks.append(95)
marks.append(100)

print(marks)
```
