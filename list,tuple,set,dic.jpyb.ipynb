```python
#tuple in python 
#tuple is a collection of multiple values that is ordered and cannot be changed after creation 
student=("Kundana",98,"python")

print(student[0])
```

```python
#acess values in a tuple
student=("Kundana",18,90)

print(student[0])
```

```python
#immutable nature of tuple
student=("Kundana",18,85,5)

student[2]=22
#this gives an error because we cannot modify a tuple
```

```python
#tuple are immutable meaning they vannot be changed after they created.ther are defined as
numbers=(10,20,20,30,20)

print(numbers.count(20))
```

```python
numbers=(10,20,30,40)

print(numbers.index(30))
```

```python
numbers=(10,20,30,40,50)

print(len(numbers))
print(max(numbers))
print(min(numbers))
print(sum(numbers))
```

```python
#set in python
#set is a collection of unique values that is unordered and mutable
numbers={10,20,30,20,10,40}

print(numbers)
```

```python
#add values to a set
subjects={"python","java"}

subjects.add("SQL")

print(subjects)
```

```python
#remove values from a set
subjects.remove("java")

print(subjects)
```

```python
#sets do nor allow the duplicate values
number={1,2,3,4,4,5,5,78,8,9,}

print(number)
 
```

```python
#dictionaries in python 
#dictionary is a collection of key-values pairs that is unorded
student={
    "name":"Kundana",
    "age":18,
    "course":"python"
}
print(student)
```

```python
#acess elements in dic
print(student["name"])
print(student["age"])
print(student["course"])
```

```python
#add new data ti a dictinary
student["city"]="tirupati"

print(student)
```

```python
#

print(student.keys())

print(student.values())

print(student.items())

print(student.get("name"))

student.update({"age":18})

print(student.pop("age"))

student.popitem()

print(student)

student.setdefault("age",18)

print(student)
```
